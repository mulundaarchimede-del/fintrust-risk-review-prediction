```python
"""
FinTrust - Week 1 Data Understanding Script
Data Science track - AnalystLab Africa

Purpose: quick exploratory checks on the two synthetic datasets
(structure, types, missing values, target balance, and a few
cross-tabulations used to ground the Week 1 hypotheses).

Run from the folder containing the CSV files.
"""

import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)

# ---------------------------------------------------------------
# 1. Load the data
# ---------------------------------------------------------------
cust = pd.read_csv("FinTrust_Customer_Data - FinTrust_Customer_Data.csv")
trans = pd.read_csv("FinTrust_Transaction_Data - FinTrust_Transaction_Data.csv")

# ---------------------------------------------------------------
# 2. Basic structure: shape, dtypes, missing values, uniques
# ---------------------------------------------------------------
print("=== CUSTOMER DATA ===")
print("Shape:", cust.shape)
print("\nColumn types:\n", cust.dtypes)
print("\nMissing values per column:\n", cust.isnull().sum())
print("\nUnique values per column:\n", cust.nunique())

print("\n=== TRANSACTION DATA ===")
print("Shape:", trans.shape)
print("\nColumn types:\n", trans.dtypes)
print("\nMissing values per column:\n", trans.isnull().sum())
print("\nUnique values per column:\n", trans.nunique())
```

```python

# ---------------------------------------------------------------
# 3. Look at the actual categories behind each categorical column
#    (nunique() only gives a count - this shows what the values are)
# ---------------------------------------------------------------
print("\nGender:", cust["Gender"].unique())
print("Customer_Segment:", cust["Customer_Segment"].unique())
print("Account_Type:", cust["Account_Type"].unique())
print("Monthly_Income_Band:", cust["Monthly_Income_Band"].unique())
print("Preferred_Channel:", cust["Preferred_Channel"].unique())
print("Account_Status:", cust["Account_Status"].unique())
print("City:", cust["City"].unique())

print("\nTransaction_Type:", trans["Transaction_Type"].unique())
print("Channel:", trans["Channel"].unique())
print("Device_Type:", trans["Device_Type"].unique())
print("Location:", trans["Location"].unique())
print("International_Transaction:", trans["International_Transaction"].unique())
print("Transaction_Status:", trans["Transaction_Status"].unique())

```

```python
# ---------------------------------------------------------------
# 4. Target variable: Risk_Review_Flag - balance check
# ---------------------------------------------------------------
print("\nRisk_Review_Flag - counts:")
print(trans["Risk_Review_Flag"].value_counts())
print("\nRisk_Review_Flag - percentages:")
print(trans["Risk_Review_Flag"].value_counts(normalize=True) * 100)

# ---------------------------------------------------------------
# 5. Cross-tabulations: does the flag rate vary by category?
#    (this is what grounded the Week 1 hypotheses)
# ---------------------------------------------------------------
print("\nRisk vs Transaction_Status:")
print(pd.crosstab(trans["Transaction_Status"], trans["Risk_Review_Flag"]))

print("\nRisk vs Transaction_Type:")
print(pd.crosstab(trans["Transaction_Type"], trans["Risk_Review_Flag"]))

print("\nRisk vs International_Transaction:")
print(pd.crosstab(trans["International_Transaction"], trans["Risk_Review_Flag"]))

print("\nAmount_NGN stats by Risk_Review_Flag:")
print(trans.groupby("Risk_Review_Flag")["Amount_NGN"].describe())

# ---------------------------------------------------------------
# 6. Date range and referential integrity between the two tables
# ---------------------------------------------------------------
trans["dt"] = pd.to_datetime(trans["Transaction_DateTime"])
print("\nDate range:", trans["dt"].min(), "to", trans["dt"].max())

customers_without_transactions = set(cust["Customer_ID"]) - set(trans["Customer_ID"])
transactions_without_customers = set(trans["Customer_ID"]) - set(cust["Customer_ID"])
print("Customers with no transaction record:", len(customers_without_transactions))
print("Transactions with no matching customer:", len(transactions_without_customers))

# ---------------------------------------------------------------
# 7. Join customer + transaction data, then check customer-level
#    attributes against the risk flag (segment, engagement, tenure)
# ---------------------------------------------------------------
df = trans.merge(cust, on="Customer_ID", how="left")

print("\nRisk rate (%) by Customer_Segment:")
print(pd.crosstab(df["Customer_Segment"], df["Risk_Review_Flag"], normalize="index") * 100)

print("\nRisk rate (%) by Account_Status:")
print(pd.crosstab(df["Account_Status"], df["Risk_Review_Flag"], normalize="index") * 100)

print("\nRisk rate (%) by Channel:")
print(pd.crosstab(df["Channel"], df["Risk_Review_Flag"], normalize="index") * 100)

print("\nMean Digital_Engagement_Score by Risk_Review_Flag:")
print(df.groupby("Risk_Review_Flag")["Digital_Engagement_Score"].mean())

print("\nMean Tenure_Months by Risk_Review_Flag:")
print(df.groupby("Risk_Review_Flag")["Tenure_Months"].mean())

# ---------------------------------------------------------------
# 8. Data-quality checks: duplicates and invalid values
# ---------------------------------------------------------------
print("\nDuplicate Transaction_ID:", trans["Transaction_ID"].duplicated().sum())
print("Duplicate Customer_ID:", cust["Customer_ID"].duplicated().sum())
print("Transactions with Amount_NGN <= 0:", (trans["Amount_NGN"] <= 0).sum())
```
